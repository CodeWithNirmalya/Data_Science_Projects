1. WHICH TEAM WON THE MATCH MOST?

```python
match_wins = df['match_winner'].value_counts()
sns.barplot(y = match_wins.index,x = match_wins.values,palette = 'viridis')
plt.title("Most Matches Winner By Team",color = 'magenta')
plt.xlabel('Match Count',color = 'red')
plt.ylabel('Team Name',color = 'red')
```

Toss Decision Trends

```python
sns.countplot(x = df['toss_decision'],palette='viridis')
plt.title('Toss Decision Trend',color = 'red')
```

Toss Winner VS Match Winner

```python
result = df[df['toss_winner'] == df['match_winner']]['match_id'].count()
print(f'The count of both match and toss winner in this year is: {result}')
percentage = (result*100)/df.shape[0]
print(f'The percentage of Toss winner and Match winner is {percentage.round(2)}%')
```

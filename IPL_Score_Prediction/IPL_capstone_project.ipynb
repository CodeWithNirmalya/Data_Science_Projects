1. WHICH TEAM WON THE MATCH MOST?

```python
match_wins = df['match_winner'].value_counts()
sns.barplot(y = match_wins.index,x = match_wins.values,palette = 'viridis')
plt.title("Most Matches Winner By Team",color = 'magenta')
plt.xlabel('Match Count',color = 'red')
plt.ylabel('Team Name',color = 'red')
```

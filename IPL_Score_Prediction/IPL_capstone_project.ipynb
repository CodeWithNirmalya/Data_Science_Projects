1. WHICH TEAM WON THE MATCH MOST?

```python
match_wins = df['match_winner'].value_counts()
sns.barplot(y = match_wins.index,x = match_wins.values,palette = 'viridis')
plt.title("Most Matches Winner By Team",color = 'magenta')
plt.xlabel('Match Count',color = 'red')
plt.ylabel('Team Name',color = 'red')
```

Toss Decision Trends

```python
sns.countplot(x = df['toss_decision'],palette='viridis')
plt.title('Toss Decision Trend',color = 'red')
```

Toss Winner VS Match Winner

```python
result = df[df['toss_winner'] == df['match_winner']]['match_id'].count()
print(f'The count of both match and toss winner in this year is: {result}')
percentage = (result*100)/df.shape[0]
print(f'The percentage of Toss winner and Match winner is {percentage.round(2)}%')
```

```python
sns.countplot(x=df['won_by'],palette='icefire')
plt.title('Team wining trends',color = 'blue')
```

Added count in this plot

```python
ax = sns.countplot(x='won_by', data=df, palette='icefire')

# Add count value above each bar
for p in ax.patches:
    ax.text(
        p.get_x() + p.get_width()/2,      # center of bar
        p.get_height() + 2,               # slightly above bar
        int(p.get_height()),              # count value
        ha='center',
        fontsize=12
    )

plt.show()
```

USING PLOTLY

```python
import plotly.express as px
from plotly.offline import iplot

counts = df['won_by'].value_counts().reset_index()
counts.columns = ['won_by', 'count']

fig = px.bar(
    counts,
    x='won_by',
    y='count',
    text='count'      # Shows values on bars
)

fig.update_traces(textposition='outside')
iplot(fig)
```
